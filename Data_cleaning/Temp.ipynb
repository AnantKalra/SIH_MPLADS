```python
# 1. Drop "Grand Total" Rows from all datasets
print("Cleaning Grand Total rows...")
for name, df in zip(['Expenditure', 'Completed', 'Recommended', 'Sanctioned'], 
                    [Expenditure, Completed, Recommended, Sanctioned]):
    if len(df) > 0 and df.iloc[-1].astype(str).str.contains('Grand Total').any():
        df.drop(df.tail(1).index, inplace=True)
        print(f"Dropped Grand Total from {name}")
```

```python
# 2. Standardize Expenditure Dataset keys before join
Expenditure.rename(columns={"Work ID": "Work_Id"}, inplace=True)

# Extract MP_ID, Financial_Year, and true numerical Work_Id
if 'Work_Id' in Expenditure.columns and Expenditure['Work_Id'].str.contains('/', na=False).any():
    Expenditure[["MP_ID", "Financial_Year"]] = Expenditure["Work_Id"].str.split("/", expand=True).iloc[:, 1:3]
    Expenditure['Work_Id'] = Expenditure['Work_Id'].str.split("/").str[3]
    Expenditure["MP_ID"] = Expenditure["MP_ID"].str.strip()
    
Expenditure[['Work_Id', 'MP_ID', 'Financial_Year', 'Fund Disbursed Amount ( ₹ )']].head(3)
```

```python
# 3. Aggregate Expenditure to fix 1-to-Many Relationship
# So that we have 1 row per Work, which is required for tabular ML

# Ensure proper types
Expenditure['Expenditure Date'] = pd.to_datetime(Expenditure['Expenditure Date'], format="%d-%b-%Y", errors='coerce')
# Convert everything to string first, replace commas, then convert back to numeric
Expenditure['Fund Disbursed Amount ( ₹ )'] = pd.to_numeric(
    Expenditure['Fund Disbursed Amount ( ₹ )'].astype(str).str.replace(',', ''), 
    errors='coerce'
)

# Aggregate the payments per Work_Id
Expenditure_Agg = Expenditure.groupby('Work_Id').agg(
    Total_Fund_Disbursed=('Fund Disbursed Amount ( ₹ )', 'sum'),
    Payment_Count=('Work_Id', 'count'),
    Latest_Payment_Date=('Expenditure Date', 'max')
).reset_index()

Expenditure_Agg.head()
```

```python
# 4. Check for True Duplicates
print("Duplicates in Completed: ", Completed[Completed['Work_Id'].notna()].duplicated(subset=['Work_Id']).sum())
print("Duplicates in Recommended: ", Recommended[Recommended['Work_Id'].notna()].duplicated(subset=['Work_Id']).sum())
print("Duplicates in Sanctioned: ", Sanctioned[Sanctioned['Work_Id'].notna()].duplicated(subset=['Work_Id']).sum())
print("Duplicates in Expenditure_Agg: ", Expenditure_Agg[Expenditure_Agg['Work_Id'].notna()].duplicated(subset=['Work_Id']).sum())

# If duplicates exist (e.g. data pipeline error), you can remove them using:
# Completed = Completed.drop_duplicates(subset=['Work_Id'], keep='first')
```

```python
# 5. Clean Null Work_Ids from RIGHT tables to avoid Cartesian Explosions
# We keep nulls in 'Recommended' (our left table) but drop them from the joining tables
Sanctioned = Sanctioned.dropna(subset=['Work_Id']).copy()
Completed = Completed.dropna(subset=['Work_Id']).copy()
Expenditure_Agg = Expenditure_Agg.dropna(subset=['Work_Id']).copy()

# 6. Prevent Column Duplicate Chaos (e.g. State_x, State_y)
# We drop general metadata from the right tables because 'Recommended' already has it.
cols_to_drop = [
    "State", "IDA", "Hon'ble Members of Parliament", "Constituency", 
    "Work category", "Work Category", "Work description", "Work Description", 
    "Recommended date", "Work", "MP_ID", "Financial_Year"
]

Sanctioned_clean = Sanctioned.drop(columns=[col for col in cols_to_drop if col in Sanctioned.columns])
Completed_clean = Completed.drop(columns=[col for col in cols_to_drop if col in Completed.columns])
```

```python
# 7. Perform the Left Joins (The Master Merge)
print(f"Initial Recommended Records: {len(Recommended)}")

# Step 1: Recommended + Sanctioned
# We use suffixes just in case there are other overlap columns (like Sanction Date)
Master_Data = Recommended.merge(Sanctioned_clean, on='Work_Id', how='left', suffixes=('', '_Sanc'))

# Step 2: Result + Completed
Master_Data = Master_Data.merge(Completed_clean, on='Work_Id', how='left', suffixes=('', '_Comp'))

# Step 3: Result + Expenditure
Master_Data = Master_Data.merge(Expenditure_Agg, on='Work_Id', how='left')

print(f"Final combined Records: {len(Master_Data)}")
Master_Data.head(3)
```

```python
Master_Data["is_Work_description"] = Master_Data["Work description"].notna().map({True: "Yes", False: "No"})
```

```python
import numpy as np

# The condition: "Completion Date" is NOT null AND "Image" IS null
missing_image_condition = Master_Data["Completion Date"].notna() & Master_Data["Image"].isna()

# np.where(condition, value_if_true, value_if_false)
Master_Data["Completed_and_images"] = np.where(missing_image_condition, "No", "Yes")
```

```python
Master_Data["District_IDA"] = Master_Data["IDA"].str.extract(r"^([^(]+)")[0].str.strip()
Master_Data
```

```python
# 8. Fix Work_Id column (Extract numeric ID and move appended text)
print("Fixing 4,535 corrupted Work_Ids...")

# Identify rows with a dash in Work_Id
mask = Master_Data['Work_Id'].astype(str).str.contains('-', na=False)

# Extract the text appended after the dash
appended_text = Master_Data.loc[mask, 'Work_Id'].str.split('-', n=1).str[1].str.strip()

# Create 'Work' column if it doesn't already exist
if 'Work' not in Master_Data.columns:
    Master_Data['Work'] = None

# Move text to 'Work' column ONLY where it is currently missing (isna)
missing_work_mask = mask & Master_Data['Work'].isna()
Master_Data.loc[missing_work_mask, 'Work'] = appended_text[Master_Data['Work'].isna()]

# Extract the purely numeric ID before the dash to fix the Work_Id column
Master_Data.loc[mask, 'Work_Id'] = Master_Data.loc[mask, 'Work_Id'].str.split('-', n=1).str[0].str.strip()

print(f"Fixed {mask.sum()} rows successfully!")
Master_Data[mask].head(3)
```
